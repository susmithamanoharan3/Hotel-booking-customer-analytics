```python
q1 = df['adr'].quantile(0.25)
q3 = df['adr'].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

outliers = df[(df['adr'] < lower_bound) | (df['adr'] > upper_bound)]

print("Outlier count:", outliers.shape[0])
```

```python
outlier_percentage = (outliers.shape[0] / df.shape[0]) * 100

print("Outlier Percentage:", outlier_percentage)
```

```python
q1 = df['lead_time'].quantile(0.25)
q3 = df['lead_time'].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

outliers_lead = df[(df['lead_time'] < lower_bound) | (df['lead_time'] > upper_bound)]

print("Lead Time Outliers:", outliers_lead.shape[0])
```

```python
q1 = df['stays_in_weekend_nights'].quantile(0.25)
q3 = df['stays_in_weekend_nights'].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

outliers_weekend = df[
    (df['stays_in_weekend_nights'] < lower_bound) |
    (df['stays_in_weekend_nights'] > upper_bound)
]

print("Weekend Nights Outliers:", outliers_weekend.shape[0])
```

```python
q1 = df['stays_in_week_nights'].quantile(0.25)
q3 = df['stays_in_week_nights'].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

outliers_week = df[
    (df['stays_in_week_nights'] < lower_bound) |
    (df['stays_in_week_nights'] > upper_bound)
]

print("Week Nights Outliers:", outliers_week.shape[0])
```
